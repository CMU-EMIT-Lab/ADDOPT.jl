```julia
using Plots
using LinearAlgebra
```

```julia
lnA = 8.845
n = 0.358
Q = 56.277e-3
R = 8.3144598e-3
dt = 1.0#50e-3
```

```julia
original_step = (y, T) -> 1 - exp(-((-log(1 - y))^(1 / n) + exp(lnA - Q / R / T) * dt)^n)
single_log_step = (ỹ, T) -> (ỹ^(1 / n) + exp(lnA - Q / R / T) * dt)^n
double_log_step = (ŷ, T) -> n * log(exp(ŷ / n) + exp(lnA - Q / R / T) * dt)
```

```julia
T = 300.0e-3:50.0e-3:1000.0e-3 # kK
y = 0.05:0.05:0.95 
ỹ = @. -log(1 - y);
ŷ = @. log(ỹ);
```

```julia
plot(y, T, original_step, st=:surface, xlabel="y", ylabel="T (kK)", camera=(-45, 45))
```

```julia
contour(y, T, original_step, xlabel="y", ylabel="T (kK)")
```

```julia
plot(ỹ, T, single_log_step, st=:surface, xlabel="ỹ", ylabel="T (kK)", camera=(-45, 45))
```

```julia
contour(ỹ, T, single_log_step, xlabel="ỹ", ylabel="T (kK)")
```

```julia
plot(ŷ, T, double_log_step, st=:surface, xlabel="ŷ", ylabel="T (kK)", camera=(-45, 45))
```

```julia
contour(ŷ, T, double_log_step, xlabel="ŷ", ylabel="T (kK)")
```
